```python
df.dropna(inplace=True)
```

```python
numeric_df=df.select_dtypes(include="number")
result=pd.DataFrame({
    "Mean":numeric_df.mean(),
    "Median":numeric_df.median(),
    "Mode":numeric_df.mode().iloc[0]  
})
print(result.round(3))
```

```python
result1=pd.DataFrame({
    "variance":numeric_df.var(),
    "standard Deviation":numeric_df.std()
})
print(result1)
```

```python
result3 = pd.DataFrame({
    "Q1(25%)": numeric_df.quantile(0.25),
    "Q2(median)": numeric_df.quantile(0.50),
    "Q3(75%)": numeric_df.quantile(0.75)
})
result3["IQR"]=result3["Q3(75%)"]-result3["Q1(25%)"]
print(result3)
```

```python
result4 = pd.DataFrame({
    "Skewness": numeric_df.skew()
})
print(result4.round(3))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns 
column = numeric_df.columns[0]

sns.histplot(numeric_df[column],kde=True)

plt.title(f"Skewness Distribution of [column]")
plt.xlabel(column)
plt.ylabel("Frequency")
plt.show()

print("Skewness:",numeric_df[column].skew())
```

```python
result5 = pd.DataFrame({
    "kurtosis": numeric_df.kurt()
})
print(result5.round(3))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns 
column = numeric_df.columns[2]

sns.histplot(numeric_df[column], kde=True)

plt.title(f"kurtosis Distribution of [column]")
plt.xlabel(column)
plt.ylabel("Frequency")
plt.show()

print("Skewness:",numeric_df[column].kurt())
```

```python
Q1 = numeric_df.quantile(0.25)
Q3 = numeric_df.quantile(0.75)

IQR = Q3 - Q1

lower_unit = Q1 - 1.5 * IQR
upper_unit = Q3 - 1.5 * IQR

outliers = (
    (numeric_df < lower_unit)|
    (numeric_df < upper_unit)
)
print("Lower Bound:\n", lower_unit)
print("Upper Bound:\n", upper_unit)
print("Outliers in each column:\n", numeric_df[outliers].count())
```

```python
mean_values = numeric_df.mean()
std_values = numeric_df.std()

result6 = pd.DataFrame({
    "Mean": mean_values,
    "Standard Deviation": std_values,
    "CV (%)": (std_values / mean_values)* 100
})

print(result6.round(3))
```
